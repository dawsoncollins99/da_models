```python
import numpy as np
from da_models.lorenz63 import rhs
from da_models.integrators import euler_maruyama
```

**Forward-Euler vs. Euler-Maruyama**

```python
def forward_euler(rhs, x0, t_span, dt, rhs_kwargs=None):
    rhs_kwargs = rhs_kwargs or {}
    t0, tf = t_span
    n_steps = int(round((tf - t0) / dt))
    t = t0 + dt * np.arange(n_steps + 1)
    x0 = np.asarray(x0, dtype=float)
    X = np.zeros((n_steps + 1, x0.shape[0]))
    X[0] = x0
    for n in range(n_steps):
        X[n + 1] = X[n] + dt * rhs(t[n], X[n], **rhs_kwargs)
    return t, X

t_fe, X_fe = forward_euler(rhs, [1.0, 1.0, 1.0], (0, 5), dt=0.01)
t_em, X_em = euler_maruyama(rhs, [1.0, 1.0, 1.0], (0, 5), dt=0.01, q=0.0, seed=0)

print(np.allclose(X_fe, X_em))
```

**Euler-Maruyama with the Same Seed**

```python
t1, X1 = euler_maruyama(rhs, [1.0, 1.0, 1.0], (0, 5), dt=0.01, q=0.5, seed=42)
t2, X2 = euler_maruyama(rhs, [1.0, 1.0, 1.0], (0, 5), dt=0.01, q=0.5, seed=42)

print(np.array_equal(X1, X2))
```

**Observations**

EM with $q=0$ does indeed yield the same thing as FE. Additionally, EM gives the same output for the same seed
